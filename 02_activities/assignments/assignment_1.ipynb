## Question One: First Duplicate in List
**Description**  
Given a list of integers, return the **first value that appears more than once**. If there are multiple duplicates, return the one that appears **first** in the list. If no duplicate exists, return `-1`.

**Examples**
```python
Input: nums = [3, 1, 4, 2, 5, 1, 6]
Output: 1
```
```python
Input: nums = [7, 8, 9, 10]
Output: -1
```
```python
Input: nums = [4, 5, 6, 4, 6]
Output: 4
```

**Question 1 Starter Code**

````python
from typing import List

def first_duplicate(nums: List[int]) -> int:
    seen = set()  
    for num in nums:
        if num in seen:      # found a duplicate
            return num
        seen.add(num)        # mark number as seen
    return -1                # no duplicates found

# Example test cases
print(first_duplicate([3, 1, 4, 2, 5, 1, 6]))   
print(first_duplicate([7, 8, 9, 10]))           
print(first_duplicate([4, 5, 6, 4, 6]))         
````


-   Code the solution to your assigned problem in Python (code chunk). Note: each problem can be solved more simply if you use an abstract data type that is suitable for that problem. Using that try to find the best time and space complexity solution!


````python
from typing import List

def first_duplicate(nums: List[int]) -> int:
    
    seen = set()  # Abstract data type: set for O(1) lookups
    
    for num in nums:
        if num in seen:
            return num  
        seen.add(num)  # Add number to set
    return -1  # No duplicates found

# New example test cases

# Example 1: first duplicate is 7
nums1 = [5, 7, 3, 7, 3, 2, 1]
print(first_duplicate(nums1))  # Expected output: 7

# Example 2: no duplicates
nums2 = [10, 27, 30, 42, 50]
print(first_duplicate(nums2))  # Expected output: -1
````

This problem asks us to find the first number in a list that appears more than once. To solve it efficiently, we use a set, which allows for constant-time (O(1)) membership checks.

Time Complexity: O(n)
Each element is checked once, and set lookups are O(1) on average.
The loop runs once per element → n iterations.
num in seen is a set lookup, which is O(1) on average (constant time).
seen.add(num) is also O(1).
Total Time Complexity = O(n)
→ The algorithm’s runtime increases linearly with the number of elements in the list.

Space Complexity: O(n)
In the worst case (no duplicates), we store all elements in the set. The memory used by the program grows linearly with the input size.


-   Explain why your solution works


The solution works because it checks each number in the list only once while keeping track of the numbers that have already been seen. As the loop goes through each element; the program looks to see if the current number is already in the seen set.
If it is, that means this number has appeared before — so it is the first duplicate, and we return it immediately.

If it is not in the set, we add it to seen and continue.

Because we go through the list from left to right, the first time we find a repeat, it must be the earliest duplicate in the list. If the loop finishes and no duplicates are found, the function correctly returns -1. Using a set ensures the lookup and insertion operations are very fast (O(1) time each), making the algorithm both simple and efficient.


-   Explain the problem’s time and space complexity


1. Time Complexity
Loop through the list once. We iterate over nums which has n elements → O(n).
Check and add to a set:
Checking if an element is in a set (num in seen) is O(1) on average.
Adding an element to a set (seen.add(num)) is also O(1) on average.
Overall time complexity:
O(n)×O(1)=O(n)

2. Space Complexity
We use a set called seen to store numbers that we’ve already encountered.
In the worst case, all numbers are unique, so the set will contain all n elements.
Space complexity:
O(n)
